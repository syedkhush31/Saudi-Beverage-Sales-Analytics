```python
revenue_check = (
    df["Units_Sold"] * df["Unit_Price_SAR"]
).round(2)

(revenue_check != df["Revenue_SAR"]).sum()
```

```python
cost_check = (
    df["Units_Sold"] * df["Unit_Cost_SAR"]
).round(2)

(cost_check != df["Cost_SAR"]).sum()
```

```python
profit_check = (
    df["Revenue_SAR"] - df["Cost_SAR"]
).round(2)

(profit_check != df["Profit_SAR"]).sum()
```

```python
margin_check = (
    (df["Profit_SAR"] / df["Revenue_SAR"]) * 100
).round(2)

(margin_check != df["Gross_Margin_Pct"]).sum()
```

```python
overall_margin = (
    df["Profit_SAR"].sum() / df["Revenue_SAR"].sum()
) * 100

overall_margin
```

```python
brand_performance = (
    df.groupby("Brand")
      .agg(
          Transactions=("Order_ID", "count"),
          Units_Sold=("Units_Sold", "sum"),
          Revenue_SAR=("Revenue_SAR", "sum"),
          Profit_SAR=("Profit_SAR", "sum")
      )
      .sort_values("Revenue_SAR", ascending=False)
)

brand_performance
```

```python
brand_performance["Gross_Margin_Pct"] = (
    brand_performance["Profit_SAR"] /
    brand_performance["Revenue_SAR"] * 100
).round(2)

brand_performance.sort_values("Gross_Margin_Pct", ascending=False)
```

```python
region_performance = (
    df.groupby("Region")
      .agg(
          Transactions=("Order_ID", "count"),
          Units_Sold=("Units_Sold", "sum"),
          Revenue_SAR=("Revenue_SAR", "sum"),
          Profit_SAR=("Profit_SAR", "sum")
      )
)

region_performance["Gross_Margin_Pct"] = (
    region_performance["Profit_SAR"] /
    region_performance["Revenue_SAR"] * 100
).round(2)

region_performance.sort_values("Revenue_SAR", ascending=False)
```

```python
channel_performance = (
    df.groupby("Sales_Channel")
      .agg(
          Transactions=("Order_ID", "count"),
          Units_Sold=("Units_Sold", "sum"),
          Revenue_SAR=("Revenue_SAR", "sum"),
          Profit_SAR=("Profit_SAR", "sum")
      )
)

channel_performance["Gross_Margin_Pct"] = (
    channel_performance["Profit_SAR"] /
    channel_performance["Revenue_SAR"] * 100
).round(2)

channel_performance.sort_values("Revenue_SAR", ascending=False)
```

```python
monthly_performance = (
    df.assign(Month=df["Date"].dt.to_period("M").astype(str))
      .groupby("Month")
      .agg(
          Transactions=("Order_ID", "count"),
          Units_Sold=("Units_Sold", "sum"),
          Revenue_SAR=("Revenue_SAR", "sum"),
          Profit_SAR=("Profit_SAR", "sum")
      )
)

monthly_performance["Gross_Margin_Pct"] = (
    monthly_performance["Profit_SAR"] /
    monthly_performance["Revenue_SAR"] * 100
).round(2)

monthly_performance
```

```python
kpi_summary = pd.DataFrame({
    "Metric": [
        "Total Transactions",
        "Total Units Sold",
        "Total Revenue (SAR)",
        "Total Cost (SAR)",
        "Total Profit (SAR)",
        "Overall Gross Margin (%)"
    ],
    "Value": [
        len(df),
        df["Units_Sold"].sum(),
        df["Revenue_SAR"].sum(),
        df["Cost_SAR"].sum(),
        df["Profit_SAR"].sum(),
        round(df["Profit_SAR"].sum() / df["Revenue_SAR"].sum() * 100, 2)
    ]
})

kpi_summary
```
